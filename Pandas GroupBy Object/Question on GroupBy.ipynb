```python
import pandas as pd
import numpy as np
ipl = pd.read_csv('deliveries.csv')
```

Question 1. Find the top 10 batsman in terms of runs

```python
print(ipl.groupby('batsman')['batsman_runs'].sum().sort_values(ascending=False).head(10))
```

Question 2. Find the batsman with max no of sixes

```python
six = ipl[ipl['batsman_runs'] == 6]
print(six.groupby('batsman')['batsman'].count().sort_values(ascending=False).head(1).index[0])
```

Question 3. Find batsman with most number of 4's and 6's in last 5 overs

```python
temp_df = ipl[ipl['over'] > 15]
temp_df = temp_df[(temp_df['batsman_runs'] == 4) | (temp_df['batsman_runs'] == 6)]
print(temp_df.groupby('batsman')['batsman'].count().sort_values(ascending=False).head(1).index[0])
```

Question 4. Find V Kohli's record against all teams

```python
temp_df = ipl[ipl['batsman'] == 'V Kohli']
print(temp_df.groupby('bowling_team')['batsman_runs'].sum().reset_index())
```

Question 5. Create a function that can return the highest score of any batsman

```python
def highest(batsman):
  temp_df = ipl[ipl['batsman'] == batsman]
  return temp_df.groupby('match_id')['batsman_runs'].sum().sort_values(ascending=False).head(1).values[0]

print(highest('DA Warner'))
```
