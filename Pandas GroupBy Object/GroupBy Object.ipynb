GroupBy Object --> Forms Group in data on the basis of some Columns.

There are two types of data numerical and categorical

groupby apply on categorical columns/data

```python
import pandas as pd
import numpy as np
movies = pd.read_csv('imdb-top-1000.csv')
genres = movies.groupby('Genre')
```

1. Applying builtin aggregation fuctions on groupby objects

```python
print(genres.sum())
print(genres.min())

#Question 1. find the top 3 genres by total earning
print(movies.groupby('Genre')['Gross'].sum().sort_values(ascending=False).head(3))

#Question 2. find the genre with highest avg IMDB rating
print(movies.groupby('Genre')['IMDB_Rating'].mean().sort_values(ascending=False).head(1))

#Question 3. find director with most popularity
print(movies.groupby('Director')['No_of_Votes'].sum().sort_values(ascending=False).head(1))

#Question 4. find number of movies done by each actor
print(movies.groupby('Star1')['Series_Title'].count().sort_values(ascending=False))
```

2. GroupBy Attributes and Methods

```python
#1. find total number of groups -> len
print(len(movies.groupby('Genre')))

#2. find items in each group -> size
print(movies.groupby('Genre').size())

#3. first()/last() -> nth item
print(genres.first())
print(genres.last())
print(genres.nth(6))

#4. get_group -> vs filtering
print(genres.get_group('Fantasy'))

#5. groups
print(genres.groups)

#6. describe
print(genres.describe())

#7. sample
print(genres.sample(2,replace=True))

#8. nunique
print(genres.nunique())
```

3. Aggregate Method/Function (agg) --> You can apply multiple aggregate function (sum,mean,max etc) on gropuby through pssing dictionary or list

```python
# Passing Dictionary
print(genres.agg(
    {
        'Runtime':'mean',
        'IMDB_Rating':'mean',
        'No_of_Votes':'sum',
        'Gross':'sum',
        'Metascore':'min'
    }
))

# Passing List
print(genres.agg(['min','max','sum']))

# Passing/Merge both Dictionary and list
print(genres.agg(
    {
        'Runtime':['min','mean'],
        'IMDB_Rating':'mean',
        'No_of_Votes':['sum','max'],
        'Gross':'sum',
        'Metascore':'min'
    }
))
```
