Merge --> We can join DataFrame on the basis of one common cloumn.

There are four types of joining
1. Inner Join
2. Left Join
3. Right Join
4. Outer/Full Join

```python
import pandas as pd
import numpy as np
courses = pd.read_csv('courses.csv')
students = pd.read_csv('students.csv')
nov = pd.read_csv('reg-month1.csv')
dec = pd.read_csv('reg-month2.csv')
matches = pd.read_csv('matches.csv')
delivery = pd.read_csv('deliveries.csv')
regs = pd.concat([nov,dec],ignore_index=True)
```

1. Inner Join --> It's Print Common items in both tables.

```python
print(students.merge(regs,how='inner',on='student_id'))

# Where; students = left table, regs = right table
# how = 'inner' (Which join is perform)
# on = 'student_id' (Joining perform on which column)
```

2. Left Join --> It's print complete left table data and common data in both tables.

```python
print(courses.merge(regs,how='left',on='course_id'))

# Where; courses = left table, regs = right table
# how = 'left' (Which join is perform)
# on = 'course_id' (Joining perform on which column)
```

3. Right Join --> It's print complete right table data and common data in both tables.

```python
temp_df = pd.DataFrame({
    'student_id':[26,27,28],
    'name':['Nitish','Ankit','Rahul'],
    'partner':[28,26,17]
})
students = pd.concat([students,temp_df],ignore_index=True)

print(students.merge(regs,how='right',on='student_id'))

# Where; students = left table, regs = right table
# how = 'right' (Which join is perform)
# on = 'student_id' (Joining perform on which column)
```

4. Full/Outer Join --> It's print complete data of right table, left table and common data on both tables.

```python
print(students.merge(regs,how='outer',on='student_id'))

# Where; students = left table, regs = right table
# how = 'outer' (Which join is perform)
# on = 'student_id' (Joining perform on which column)
```

5. Self Join --> Joining DataFrame with itself.

```python
students.merge(students,how='inner',left_on='partner',right_on='student_id')

# left_on for left table column
# right_on for right table column
```
