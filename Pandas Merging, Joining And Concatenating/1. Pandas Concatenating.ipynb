```python
import pandas as pd
import numpy as np
courses = pd.read_csv('courses.csv')
students = pd.read_csv('students.csv')
nov = pd.read_csv('reg-month1.csv')
dec = pd.read_csv('reg-month2.csv')
matches = pd.read_csv('matches.csv')
delivery = pd.read_csv('deliveries.csv')
```

1. pd.concat --> It's help to stack (ADD) Data/DataFrame Vertically.

```python
print(pd.concat([nov,dec]))
print(pd.concat([nov,dec]).shape)

regs = pd.concat([nov,dec],ignore_index=True)
print(regs)
```

2. Multi-index DataFrame --> key = ['1','2','3',.....]

```python
multi = pd.concat([nov,dec],keys = ['nov','dec'])
print(multi)

# Fetching Data from multi-index
print(multi.loc['nov'])
print(multi.loc['dec'])

print(multi.loc['nov',0])   # In nov index 0  value
print(multi.loc['dec',4])   # In dec index 4 value
```

We can also stack (ADD) DataFrame horizontally / Side by Side.

```python
print(pd.concat([nov,dec],axis=1))
print(pd.concat([nov,dec],axis=1).shape)
```
