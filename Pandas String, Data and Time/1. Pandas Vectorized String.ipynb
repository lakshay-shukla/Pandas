#### 2. Problem in vectorized operations in vanilla python

```python
s = ['cat','mat',None,'rat']
[i.startswith('c') for i in s]   # error --> 'NoneType' object has no attribute 'startswith'
```

#### 3. How pandas solves this issue?
String Accessor (.str) --> The pandas .str accessor allows you to perform vectorized string operations element-wise on an entire pandas series or index. And it is fast and optimized

```python
s = pd.Series(['cat','mat',None,'rat'])
# string accessor --> str
print(s.str.startswith('c'))
```

```python
# import titanic
df = pd.read_csv('titanic.csv')
print(df['Name'])
```

#### 4. Common Functions-:

```python
#1. lower/upper/capitalize/title
print(df['Name'].str.lower())   # All in lower case
print(df['Name'].str.upper())   # All in upper case
print(df['Name'].str.capitalize())   # Only First letter of full sentence is upper
print(df['Name'].str.title())   # Only first letter of all word id capital
```

```python
#2. len
print(df['Name'][df['Name'].str.len()])
print(df['Name'][df['Name'].str.len() == 82].values[0])
```

```python
#3. strip --> It's remove unnecessary Spaces.
print("                   nitish                              ".strip())
print(df['Name'].str.strip())
```

```python
#4. split -> get
df['lastname'] = df['Name'].str.split(',').str.get(0)
df.head()
```

```python
df[['title','firstname']] = df['Name'].str.split(',').str.get(1).str.strip().str.split(' ', n=1, expand=True)   # n is no. of splits
print(df.head())

print(df['title'].value_counts())
```

```python
#5. replace
df['title'] = df['title'].str.replace('Ms.','Miss.')
df['title'] = df['title'].str.replace('Mlle.','Miss.')
df['title'].value_counts()
```
