```python
import numpy as np
import pandas as pd
```

1. Creating Data Frames

```python
#1. Using Lists
student_data = [
    [100,80,10],
    [90,70,7],
    [120,100,14],
    [80,50,2]
]
student = pd.DataFrame(student_data, columns =['iq', 'marks', 'package'])
print(student)

#2. Using Dictionary
student_dict = {
    'iq' : [100,90,120,80],
    'marks' : [80,70,100,50],
    'package' : [10,7,14,2]
}
student1 = pd.DataFrame(student_dict)
print(student1)

#3. using read_csv
movies = pd.read_csv('movies.csv')
print(movies)
ipl = pd.read_csv('ipl-matches.csv')
print(ipl)
```

2. Data Frames Attributes And Methods

```python
#1. Shape --> How many no. of rows and columns in data frame
print(movies.shape)
print(ipl.shape)

#2. dtypes --> Data type of all columns
print(movies.dtypes)
print(ipl.dtypes)

#3. index --> Fetching index
print(movies.index)
print(ipl.index)

#4. columns --> Fetching all columns
print(movies.columns)
print(ipl.columns)
print(student.columns)

#5. values --> Fetching all values from data frame
print(student.values)
print(ipl.values)

#6. head and tail --> Fetching top and bottom values
print(movies.head())
print(ipl.tail())

#7. sample --> Fetching random values
print(ipl.sample())

#8. info --> High level information about data frame
print(movies.info())
print(ipl.info())

#9. describr --> Gives Mathematical Summary
print(movies.describe())
print(ipl.describe())

#10. isnull --> Missing values present or not
print(movies.isnull())   # gives true/false
print(movies.isnull().sum())   # gives value

#11. duplicated --> Check duplicate data
print(movies.duplicated())   # gives true/false
print(movies.duplicated().sum())   # gives value

#12. rename --> Rename column name
print(student.rename(columns = {'marks' : 'percent' , 'package' : 'LPA' }))
```
