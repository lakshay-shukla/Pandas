```python
import numpy as np
import pandas as pd
movies = pd.read_csv('movies.csv')
ipl = pd.read_csv('ipl-matches.csv')
student_dict = {
    'iq' : [100,90,120,80],
    'marks' : [80,70,100,50],
    'package' : [10,7,14,2]
}
student1 = pd.DataFrame(student_dict)
```

1. DataFrame Mathematical Methods --> All Series Methods are applicable on dataframe

Axis --> Row = 1 and Column = 0

```python
#1. count --> Total no. of rows but it's not count missing values.
print(movies.count())

#2. sum/product --> By default it's apply on all columns/columnwise.
print(student1.sum())   # by default columnwise
print(student1.sum(axis=1))

#3. mean/median/mode/std/var -->
print(student1.mean())
print(student1.median())
print(student1.mode())

#4. min/max
print(student1.min())
print(student1.max())

#5. describe --> It's provides a summary of multiple things
print(student1.describe())
```

2. Selcting Columns from a DataFrame

```python
#1. Single Column
print(movies['title_x'])   # title_x is Column name
print(ipl['Venue'])

#2. Multiple Columns
print(movies[['year_of_release','actors','title_x']])  # You Can Decide order of Fetching
print(ipl[['Team1','Team2','WinningTeam']])
```

3. Selecting Rows from a DataFrame --> There are two methods for Selecting Rows

i). iloc --> It's Searches using index Positions  (only first range included)

ii). loc --> It's Searches using Index Lables  (Both range included)

```python
student_dict = {
    'name':['nitish','ankit','rupesh','rishabh','amit','ankita'],
    'iq':[100,90,120,80,0,0],
    'marks':[80,70,100,50,0,0],
    'package':[10,7,14,2,0,0]
}
students = pd.DataFrame(student_dict)
students.set_index('name',inplace=True)

#1. iloc --> Searches using index positions
#i. Single Row
print(movies.iloc[0])

#ii. Multiple Rows
print(movies.iloc[0:5])
print(movies.iloc[5:16])
print(movies.iloc[[0,4,5]])  # Fancy Indexing

#2. loc --> Searches using index lables
print(students.loc['nitish'])
print(students.loc['nitish':'rishabh':2])
print(students.loc[['nitish','ankita','rupesh']])  # Fancy Indexing
```

4. Selecting Both Rows and Columns

```python
print(movies.iloc[0:3,0:3])   # before comma for rows and after comma for Column
print(movies.loc[0:2,'title_x':'poster_path'])   # before comma for rows and after comma for Column
```
