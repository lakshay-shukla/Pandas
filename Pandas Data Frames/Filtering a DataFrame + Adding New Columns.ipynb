```python
import numpy as np
import pandas as pd
movies = pd.read_csv('movies.csv')
ipl = pd.read_csv('ipl-matches.csv')
```

1. Filtering a DataFrame

```python
#Question 1. find all the final winners
mask = ipl['MatchNumber'] == 'Final'
new_df = ipl[mask]
print(new_df[['Season','WinningTeam']])
    # 2nd and short method Both give same output
print((ipl[ipl['MatchNumber'] == 'Final'][['Season','WinningTeam']]))
```

```python
#Question 2. how many super over finishes have occured
print(ipl[ipl['SuperOver'] == 'Y'].shape[0])
```

```python
#Question 3. how many matches has csk won in kolkata
print(ipl[(ipl['City'] == 'Kolkata') & (ipl['WinningTeam'] == 'Chennai Super Kings')].shape[0])
```

```python
#Question 4. toss winner is match winner in percentage
print((ipl[ipl['TossWinner'] == ipl['WinningTeam']].shape[0]/ipl.shape[0])*100)
```

```python
#Question 5. movies with rating higher than 8 and votes>10000
print(movies[(movies['imdb_rating'] > 8) & (movies['imdb_votes'] > 10000)].shape[0])
```

```python
#Question 6. Action movies with rating higher than 7.5
# mask1 = movies['genres'].str.split('|').apply(lambda x:'Action' in x)
mask1 = movies['genres'].str.contains('Action')
mask2 = movies['imdb_rating'] > 7.5
print(movies[mask1 & mask2])
# Both mask1 are correct but we use different funtion and 2nd mask1 is better choice
```
