#### Wide Data-:
1. Wide format is where we have a single row for every data point with multiple columns to hold the values of various attributes.
2. Example -: We have two entities (John,Christopher) and Each entity has only one row (one row for john and one row for christopher).
3. No. of Columns is more.

#### Long/Tall Data-:
1. Long format is where, for each data point we have as many rows as the number of attributes and each row contains the value of a particular attribute for a given data point.
2. Example-: We have two entities (John,Christopher) and Each entity has two or more rows (two row for john and two row for christopher).
3. No. of Rows is more.

Both Carries same information in different design/format.

#### 1. Melt-:
It's help to Convert Wide data format into long data format.

```python
import numpy as np
import pandas as pd
```

```python
#1. melt -> simple example branch
print(pd.DataFrame({'cse':[120]}).melt())

#2. melt -> branch with year
print(pd.DataFrame({'cse':[120],'ece':[100],'mech':[50]}).melt(var_name='branch',value_name='num_students'))

print(pd.DataFrame(
    {
        'branch':['cse','ece','mech'],
        '2020':[100,150,60],
        '2021':[120,130,80],
        '2022':[150,140,70]
    }
).melt(id_vars=['branch'],var_name='year',value_name='students'))
```

```python
#3. melt -> real world example
death = pd.read_csv('time_series_covid19_deaths_global.csv')
confirm = pd.read_csv('time_series_covid19_confirmed_global.csv')
print(death.head())
print(confirm.head())
```

```python
death = death.melt(id_vars=['Province/State','Country/Region','Lat','Long'],var_name='date',value_name='num_deaths')
confirm = confirm.melt(id_vars=['Province/State','Country/Region','Lat','Long'],var_name='date',value_name='num_cases')
print(death.head())
```

```python
print(confirm.merge(death,on=['Province/State','Country/Region','Lat','Long','date'])[['Country/Region','date','num_cases','num_deaths']])
```
