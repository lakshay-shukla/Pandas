```python
import numpy as np
import pandas as pd
```


### 1. MultiIndex Series


1. Can we have multiple index? Let's try

```python
index_val = [('cse',2019),('cse',2020),('cse',2021),('cse',2022),('ece',2019),('ece',2020),('ece',2021),('ece',2022)]
a = pd.Series([1,2,3,4,5,6,7,8],index=index_val)
print(a)

# The problem?
print(a['cse'])   # It gives error

# The solution -> Create MultiIndex Object , multiindex series is also known as Hierarchical Indexing
# multiple index levels within a single index
```

2. How to create multiindex object

```python
# 1. pd.MultiIndex.from_tuples()
index_val = [('cse',2019),('cse',2020),('cse',2021),('cse',2022),('ece',2019),('ece',2020),('ece',2021),('ece',2022)]
multiindex = pd.MultiIndex.from_tuples(index_val)
print(multiindex)

# 2. pd.MultiIndex.from_product()
print(pd.MultiIndex.from_product([['cse','ece'],[2019,2020,2021,2022]]))
```

3. Creating a series with multiindex object

```python
s = pd.Series([1,2,3,4,5,6,7,8],index=multiindex)
print(s)
```

4. How to fetch items from such a series

```python
print(s['cse'])
print(s['ece'])
print(s['cse',2022])
```
