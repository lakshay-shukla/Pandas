```python
import numpy as np
import pandas as pd
```

1. Series using read_csv --> Creating Series using real world data sets.

CSV --> Comma (,) separated Values

read_csv --> For read and import csv files

squeeze() --> it's convert data frames into series

```python
# with one col
subs = pd.read_csv('subs.csv').squeeze()
print(subs)

# with 2 cols
vk = pd.read_csv('kohli_ipl.csv',index_col='match_no').squeeze()
print(vk)
movies = pd.read_csv('bollywood.csv',index_col='movie').squeeze()
print(movies)
```

2. Series methods

(inplace = True) --> use when we want permanent changes in original file

```python
#1. head --> To fetch top values
print(subs.head())  # By Default fetch top 5
print(vk.head(3))   # fetch top 3

#2. tail --> To fetch bottom values
print(vk.tail())  # By Default fetch bottom 5
print(vk.tail(10))  # fetch bottom 10

#3. sample --> Randomly fetch only 1 row or value
print(movies.sample())
print(movies.sample(5))

#4. value_counts --> Frequency of data
print(movies.value_counts())

#5. sort_values --> Sort the values
print(vk.sort_values())  # Ascending Order
print(vk.sort_values(ascending=False))  # Descending order
print(vk.sort_values(ascending=False).head(1).values[0])  # Chaining Method
print(vk.sort_values(inplace=True))

#6. sort_index --> Sort the index
print(movies.sort_index())  # Ascending order
print(movies.sort_index(ascending=False))   # Descending order
print(movies.sort_index(ascending=False,inplace=True))
```

3. Series Maths Method

```python
#1. count --> Total no. of items but it's not count missing values.
print(vk.count())

#2. sum/product -->
print(subs.sum())
print(subs.product())

#3. mean/median/mode/std/vsr -->
print(subs.mean())   # Average
print(vk.median())  # Mid Value
print(movies.mode())   # Most Repeated Values
print(subs.std())
print(vk.var())

#4. min/max
print(subs.min())
print(subs.max())

#5. describe --> It's provides a summary of multiple things
print(subs.describe())
```
