```python
import numpy as np
import pandas as pd
```

```python
subs = pd.read_csv('subs.csv').squeeze()
vk = pd.read_csv('kohli_ipl.csv',index_col='match_no').squeeze()
movies = pd.read_csv('bollywood.csv',index_col='movie').squeeze()
marks = [67,57,89,100]
subjects = ['maths','english','science','hindi']
print(pd.Series(marks,index=subjects))
```

1. Series with Python Functionalities

```python
#1. len/type/dir/sorted/max/min
print(len(subs))  # 365
print(type(subs))   # type Series
print(dir(subs))    # It gives Attributes and methods of series class
print(sorted(subs))  # It gives Sorted list
print(min(subs))    # 33
print(max(subs))   # 396

#2. type conversion
print(list(marks))

#3. membership operator  (in , not in)
print('2 States (2014 film)' in movies)
print('Alia Bhatt' in movies.values)

#4. looping
for i in movies.index:
  print(i)

#5. Arithmetic Operators(Broadcasting)  --> +,-,*,**,/,//,%
print(100 + subs)

#6. Relational Operators  --> >,<,<=,>=,==,!=
print(vk >= 50)
```

2. Boolean Indexing on Series

```python
#1. Find no of 50's and 100's scored by kohli
print(vk[vk >= 50].size)

# find number of ducks
print(vk[vk == 0].size)

# Count number of day when I had more than 200 subs a day
print(subs[subs > 200].size)

# find actors who have done more than 20 movies
num_movies = movies.value_counts()
print(num_movies[num_movies > 20])
```

3. Plotting Graph on Series

```python
#1. Graph 1
print(subs.plot())
```
