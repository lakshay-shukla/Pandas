```python
import numpy as np
import pandas as pd
```

```python
subs = pd.read_csv('subs.csv').squeeze()
vk = pd.read_csv('kohli_ipl.csv',index_col='match_no').squeeze()
movies = pd.read_csv('bollywood.csv',index_col='movie').squeeze()
runs = [13,24,56,78,100]
print(pd.Series(runs))
marks = [67,57,89,100]
subjects = ['maths','english','science','hindi']
print(pd.Series(marks,index=subjects))
```

1. Series Indexing --> Fetching the item

```python
x = pd.Series([12,13,14,35,46,57,58,79,9])
#1. integer indexing --> Positive indexing
print(x[0])
print(x[1])

#2. Negative indexing --> Series not work on negative indexing
```

2. Slicing --> Fetch multiple items

```python
print(vk[5:16])   # Positive slicing
print(vk[-5:])   # Negative Slicing
print(movies[-5:])
print(movies[::2])
```

3. Fancy Indexing And Indexing with labels

```python
# fancy indexing
print(vk[[1,3,4,5]])

# indexing with labels -> fancy indexing
print(movies['2 States (2014 film)'])
```

4. Editing Series

```python
#1. using indexing
marks[1] = 100
print(marks)

#2. what if an index does not exist
marks['evs'] = 100
print(marks)

#3. slicing
runs[2:4] = [100,100]
print(runs)

# fancy indexing
runs[[0,3,4]] = [0,0,0]
print(runs)

# using index label
movies['2 States (2014 film)'] = 'Alia Bhatt'
print(movies)
```
