What is Pandas --> Pandas is a fast, powerful, flexible and easy to use open source data analysis and manipulation tool, built on top of the Python programming language.

Pandas Series --> A Pandas Series is like a column in a table. It is a 1-D array holding data of any type.

```python
import numpy as np
import pandas as pd
```

1. Series from lists --> Series object have two things (index + values)

```python
#1. String
country = ['India','Pakistan','USA','Nepal','Srilanka']
print(pd.Series(country))

#2. Integers
runs = [13,24,56,78,100]
print(pd.Series(runs))

#3. Custom Index
marks = [67,57,89,100]
subjects = ['maths','english','science','hindi']
print(pd.Series(marks,index=subjects))

#4. Setting a Name --> You can give name to your series object
marks = pd.Series(marks,index=subjects,name='Lakshay Marks')
print(marks)
```

2. Series from Dictionary

```python
marks = {
    'maths':67,
    'english':57,
    'science':89,
    'hindi':100
}
marks_series = pd.Series(marks,name='Lakshay Marks')
print(marks_series)
```

3. Series Attributes

```python
#1. size --> Items in the series and it's count missing values also.
print(marks_series.size)

#2. dtype --> Data type of items
print(marks_series.dtype)

#3. name --> Name of series object
print(marks_series.name)

#4. is_unique --> Series items is unique or not
print(marks_series.is_unique)
print(pd.Series([1,1,2,3,4,5]).is_unique)

#5. index --> This print all indexes
print(marks_series.index)

#6. values --> To extract all values
print(marks_series.values)
```
