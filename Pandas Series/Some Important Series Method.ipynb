```python
import numpy as np
import pandas as pd
subs = pd.read_csv('subs.csv').squeeze()
vk = pd.read_csv('kohli_ipl.csv',index_col='match_no').squeeze()
movies = pd.read_csv('bollywood.csv',index_col='movie').squeeze()
```

1. astype

```python
import sys
print(sys.getsizeof(vk))
print(sys.getsizeof(vk.astype('int16')))
```

2. between --> Values exist in a given range or not and both terms are included.

```python
print(vk.between(51,99))
print(vk[vk.between(51,99)].size)
```

3. clip --> clip in a particular range ;

if number is < 100 = 100 or number is > 200 = 200

```python
print(subs.clip(100,200))
```

4. drop_duplicates

```python
temp = pd.Series([1,1,2,2,3,3,4,4])
print(temp)
print(temp.drop_duplicates())  # by default it removes last no.
print(temp.drop_duplicates(keep='last'))   # it removes first no.

# To Check duplicate values
print(temp.duplicated())  # gives true and false
print(temp.duplicated().sum())  # 4   it tells no. of duplicates value
```

5. isnull --> It's check value is missing or not

```python
print(temp.isnull())  # gives true and false
print(temp.isnull().sum())  # gives no. of missing values (0)
print(vk.isnull().sum()) # 0
```

6. dropna --> To remove missing values

```python
print(temp.dropna())
```

7. fillna --> To fill the missing values with any value you want.

```python
print(temp.fillna(0))  # if missing vales is present then missing values replace with 0
print(temp.fillna(temp.mean()))
```

8. isin --> It's check value is exist or not.

```python
print(vk.isin([49,99]))  # gives true and false
print(vk[vk.isin([49,99])])
```

9. apply --> It's help to apply custom logic on series.

```python
print(movies.apply(lambda x:x.split()[0].upper()))
print(subs.apply(lambda x:'good day' if x > subs.mean() else 'bad day'))
```

10. copy --> It's make a copy/replica of series then it's changes perform on copy not in original series.

head and tail gives view of the data only

```python
print(vk)
new = vk.head()
new[1] = 1
print(new)
new = vk.head().copy()
new[1] = 100
print(new)
print(vk)
```
